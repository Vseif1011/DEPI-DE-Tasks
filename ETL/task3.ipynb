```python
import pyodbc
from google import genai
import pandas as pd
```

```python
conn = pyodbc.connect(
    "DRIVER={ODBC Driver 18 for SQL Server};"
    "SERVER=localhost;"
    "DATABASE=Company_SD;"
    "Trusted_Connection=yes;"
    "TrustServerCertificate=yes;"
)
```

```python
import os

client = genai.Client(api_key=os.environ["GEMINI_API_KEY"])
def ask_model(q):
    interaction = client.interactions.create(
        model="gemini-3.8-flash",
        input= f'''
        Database Schema: 
        
        Departments(
        Dname nvarchar(50) NULL,
        Dnum int PRIMARY KEY NOT NULL,
        MGRSSN int FOREIGN KEY NULL,
        MGRStart_Date datetime NULL
    )

    Dependent(
        ESSN int PRIMARY KEY, FOREIGN KEY NOT NULL,
        Dependent_name nvarchar(255) PRIMARY KEY NOT NULL,
        Sex nvarchar(255) NULL,
        Bdate datetime NULL
    )

    Employee(
        Fname nvarchar(50) NULL,
        Lname nvarchar(50) NULL,
        SSN int PRIMARY KEY NOT NULL,
        Bdate datetime NULL,
        Address nvarchar(50) NULL,
        Sex nvarchar(50) NULL,
        Salary int NULL,
        Superssn int FOREIGN KEY NULL,
        Dno int FOREIGN KEY NULL
    )

    Project(
        Pname nvarchar(50) NULL,
        Pnumber int PRIMARY KEY NOT NUL,
        Plocation nvarchar(50) NULL, 
        City nvarchar(50) NULL,
        Dnum int FORIGEN KEY NULL
    )

    WORKS_FOR(
        ESSn int PRIMARY KEY FORGEIN KEY NOT NULL,
        PNO int PRIMARY KEY FORGEIN KEY NOT NULL,
        Hours int NULL
    )
        
    Answer the following question and return a pure sql query without new line laterals or any extras: {q}

        '''
    )
    query = interaction.output_text 
    return query

```

```python
Done = False
df = None 
question = input("Enter a question: ")
while(not Done):
    try: 
        query = ask_model(question)
        df = pd.read_sql_query(query,conn)
        Done = True
    except Exception: 
        continue
df
```
