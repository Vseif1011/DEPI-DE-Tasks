### 1. Extract Data

```python
import requests 

Cities = ["Cairo",
    "Giza",
    "Alexandria",
    "Port Said",
    "Suez",
    "Ismailia",
    "Mansoura",
    "Tanta",
    "Luxor",
    "Aswan"]

listOfCities = []

for city in Cities : 
    url = "https://api.openweathermap.org/data/2.5/weather"
    params = {
        "appid": "6a13e61e853763292101962a026aa3e8",
        "q": city
    }
    response = requests.get(url, params=params)
    data = response.json()
    listOfCities.append(data)
    print(data)
    print('---')

```

```python
import pandas as pd
```

```python
df = pd.DataFrame(listOfCities)
df
```

```python
df.drop(columns=['Latitude'])
```

```python
CitiesData = []

for country in listOfCities:

    CityName = country['name']
    CountryName = country['sys']['country']
    Latitude = country['coord']['lat']
    Longitude = country['coord']['lon']
    Temperature = country['main']['temp']
    Pressure = country['main']['pressure']
    Humidity = country['main']['humidity']
    Windspeed = country['wind']['speed']

    CountryJson = {
        'City': CityName,
        'Country': CountryName,
        'Latitude': Latitude,
        'Longitude': Longitude,
        'Temperature': Temperature,
        'Pressure': Pressure,
        'Humidity': Humidity,
        'Wind Speed': Windspeed
    }

    CitiesData.append(CountryJson)
    
```

```python
df = pd.DataFrame(CitiesData)
df
```

```python
def classify(city):
    if city == 'Cairo' or city == 'Alexandria':
        return 'Upper'
    else:
        return 'Lower'
df['Region']=df['City'].apply(classify)
df
```

```python
df['Temperature'] = df['Temperature'] - 273.15
df
```

```python
df.isna().sum()
```

```python
def classifyTemp(temp): 
    if temp <= 25: 
        return 'Cold'
    elif 25<temp<=30:
        return 'Moderate'
    elif temp>30: 
        return 'Hot' 
df['Temperature Category'] = df['Temperature'].apply(classifyTemp)
df
```

```python
def classifyHumd(humd):
    if humd <=30: 
        return "Low"
    elif humd <=45:
        return "Moderate"
    elif humd >45: 
        return "High"
df['Humidity Category'] = df['Humidity'].apply(classifyHumd)
df
```

```python
import datetime 
date = datetime.datetime.now().strftime('%Y-%m-%d')
```

```python
date = datetime.datetime.now().strftime('%Y-%m-%d')

df['Ingestion Time'] = date
df
```
