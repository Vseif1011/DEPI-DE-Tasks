```python
#3
students = ["Ahmed", "Mahmoud", "Mohamed", "Ali", "Seif", "Amr", "Omar"]
students.append("Taha")
students.pop(2)
students.sort()
print("Students in alphabetical order:")
print(students)
```

```python
#4
count = 0
for stu in students:
    if stu[0] == 'A':
        count+=1
print(f"Number of students' names starting with (A) are: {count} students.")
```

```python
#5
products = {
    "pizza": 10,
    "shawrma": 5,
    "meal": 15,
    "cola": 8
}

print("Available Products:")
for product, price in products.items():
    print(f"{product}: ${price}")

total = 0

while True:
    choice = input("Enter a product name (or type 'exit' to quit): ").lower()

    if choice == "exit":
        break

    if choice in products:
        print(f"The price of {choice} is ${products[choice]}")
        total += products[choice]
    else:
        print("Product not found")

print(f"Total amount = ${total}")
```
