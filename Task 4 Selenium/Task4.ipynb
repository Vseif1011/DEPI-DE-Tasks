```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.options import Options
import time
import pandas as pd
 
options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)
# options.add_argument("--headless=new")
options.add_argument("--window-size=1920,1080")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
```

```python
Jobs = []
driver = webdriver.Chrome(options=options)
driver.get("https://www.naukrigulf.com/data-engineer-jobs")
time.sleep(3)
Page = True
while(Page):
    time.sleep(3)
    jobPostings = driver.find_elements(By.CSS_SELECTOR,".ng-box.srp-tuple")
    for job in jobPostings:
        try:
            jobtitle = job.find_element(By.CLASS_NAME, "designation-title").text
            jobCompany = job.find_element(By.CLASS_NAME, "info-org").text
            jobLocation = job.find_element(By.CSS_SELECTOR, ".info-loc.has-logo").text
            jobExp = job.find_element(By.CSS_SELECTOR, ".info-exp.has-logo").text
            jobDesc = job.find_element(By.CLASS_NAME, "description").text

            Jobs.append({
                "Title": jobtitle,
                "Company": jobCompany,
                "Location": jobLocation,
                "Experience": jobExp,
                "Description": jobDesc
            })

        except Exception:
            continue
    try: 
        driver.find_element(By.CSS_SELECTOR,".ico.bwd").click()
    except Exception:
        Page = False
    


for j in Jobs: print(j)
```

```python
df = pd.DataFrame(Jobs)
df
```

```python
df.to_csv("naukrigulf_DE_Postings.csv")
```
