# Data Cleaning

```python
# 1- Renaming
df.rename(columns={'   Customer type':'Customer Type','   cogs':'COGS','Payment    ':'Payment Type','Unit price':'Unit Price',
                   'gross income':'Gross Income','gross margin percentage':'Gross Margin Percentage'},inplace=True) 
df.head(0)
```

```python
# 2- Handling missing values (Unit Price, Sales, and Date)

df['Date'] = df['Date'].fillna('Unknown')

df['Unit Price'] = df['Unit Price'].fillna(0.0)

df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')
df['Sales'] = df['Sales'].fillna(0.0)
        
        
df.isna().sum()
```

```python
# 3- Duplicate rows removal
df = df.drop_duplicates()
df.duplicated()
```

```python
# 4- Unnecessary Rows Removal
None
```

```python
# 5- Correct Data and Datatypes
df.info()
```

```python
df['Date'] = pd.to_datetime(df['Date'], format='%m/%d/%Y', errors='coerce')
df['Time'] = pd.to_datetime(df['Time'], format='%I:%M:%S %p')
```

```python
for i,value in enumerate(df['Gender']):
    if value == 'M':
        df.loc[i,'Gender'] = 'Male'
    elif value == 'F':
        df.loc[i,'Gender'] = 'Female'
        
df['Gender'].value_counts()
```

```python
df['Branch']=df['Branch'].str.strip()
for i,value in enumerate(df['Branch']):
    if value == 'giza':
        df.loc[i,'Branch'] = 'Giza'
    elif value == 'cairo':
        df.loc[i,'Branch'] = 'Cairo'
        
df['Branch'].value_counts()
```

```python
for i,value in enumerate(df['Payment Type']):
    if value == 'cash':
        df.loc[i,'Payment Type'] = 'Cash'
    elif value == 'credit card':
        df.loc[i,'Payment Type'] = 'Credit card'
    elif value == 'Ewallet': 
        df.loc[i,'Payment Type'] = 'E-wallet'
df['Payment Type'].value_counts()
```

```python
# 6- Sorting according to date and time
df = df.sort_values(by=['Date', 'Time'], ascending=[False, True])
```

```python
# 7- New day month and year columns 
df['Day'] = df['Date'].dt.day
df['Month'] = df['Date'].dt.strftime('%b')
df['Year'] = df['Date'].dt.year

df = df.drop(columns=['Date'])
df.head(5)
```

```python
df['Time'] = pd.to_datetime(
    df['Time'],
    format='%I:%M:%S %p'
).dt.time
```

```python
# 8- Satisfied column
df['Satisfied'] = df['Rating']>=7
df.head(10)
```

# Data Analysis

```python
# 1- Total Revenue 
totalRevenue = df['Sales'].sum()
totalRevenue
```

```python
# 2- Total revenue by city and rank them 
cityRevenue = df.groupby('City')['Sales'].sum()
cityRevenue
```

```python
cityRevenue.sort_values(ascending=False)
```

```python
# 3- Total profit by branch and rank them
totalProfitBranch = df.groupby('Branch')['Gross Income'].sum()
totalProfitBranch
```

```python
totalProfitBranch.sort_values(ascending=False)
```

```python
# 4- Total profit and revenue by category 
productCategoryAnalysis = df.groupby('Product line').agg(Revenue = ('Sales','sum'), Profit = ('Gross Income','sum'))
productCategoryAnalysis
```

```python
productCategoryAnalysis.sort_values(by=['Revenue','Profit'], ascending=[False,False])
# Food is the top performing category
```

```python
# 5- Total spending by costumer type 
costumerSpendings = df.groupby('Customer Type')['Sales'].sum().sort_values(ascending=False)
costumerSpendings
# A member spends more money
```

```python
# 6- Number of transactions by payment method 
df['Payment Type'].value_counts().sort_values(ascending=False)
# E-wallet is the most popular
```

```python
# 7- The average transaction value of the supermarket 
df['Sales'].mean()
```

```python
# 8- Satisfaction rate by branch 
satisfactionByBranch = df.groupby('Branch').agg(AverageSatisfaction = ('Satisfied','mean'), AverageRating = ('Rating','mean'))
satisfactionByBranch
# Alex branch is the winner winner chicken dinner 
```

```python
# 9- Total revenue by month and day
totalRevenueMonth = df.groupby('Month')['Sales'].sum().sort_values(ascending=False)
totalRevenueMonth
```

```python
totalRevenueDay = df.groupby('Day')['Sales'].sum().sort_values(ascending=False)
totalRevenueDay
```
